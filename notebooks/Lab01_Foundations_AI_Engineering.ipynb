## Task 3 Git Version Control

````python
from pathlib import Path

root = Path.cwd() if (Path.cwd() / ".gitignore").exists() else Path.cwd().parent
gitignore = root / ".gitignore"
content = gitignore.read_text()
print(".gitignore exists:", gitignore.exists())
print("Gitignore contains .venv:", ".venv/" in content)
print("Gitignore contains Jupyter checkpoints:", ".ipynb_checkpoints/" in content)
print("Gitignore contains .pt:", "*.pt" in content)
print("Gitignore contains .onnx:", "*.onnx" in content)
print("Gitignore contains .h5:", "*.h5" in content)
````
